```python
import requests
import pandas as pd

URL="https://gbfs.citibikenyc.com/gbfs/en/station_information.json"

response=requests.get(URL)
data=response.json()
station_list=data['data']['stations']

df=pd.DataFrame(station_list)
duplicate=df['station_id'].duplicated().sum()

print(duplicate)
print(df.shape)
print(df.head())

```
