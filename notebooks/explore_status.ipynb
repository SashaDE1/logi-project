```python
import requests
import pandas as pd

URL="https://gbfs.citibikenyc.com/gbfs/en/station_status.json"

response=requests.get(URL)
data=response.json()
stations_list=data["data"]["stations"]

df=pd.DataFrame(stations_list)
duplicate=df['station_id'].duplicated().sum()
count_station_not_work=((df['is_installed']==0) | (df['is_renting']==0)).sum()


print(duplicate)
print(count_station_not_work)
```
